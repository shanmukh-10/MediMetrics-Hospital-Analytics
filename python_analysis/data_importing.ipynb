```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# MySQL connection
url = URL.create(
    "mysql+pymysql",
    username="root",
    password="shanmukh@9848",
    host="127.0.0.1",
    port=3306,
    database="medimetrics"
)

engine = create_engine(url)

# Load cleaned appointments
appointments = pd.read_csv("../data/processed/cleaned_appointments.csv")

print("Dataset loaded:", appointments.shape)

# Import into MySQL
appointments.to_sql(
    name="appointments",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=500,
    method="multi"
)

print("Successfully imported appointments into MySQL!")
```

```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
from getpass import getpass

# MySQL connection
url = URL.create(
    "mysql+pymysql",
    username="root",
    password=getpass("Enter MySQL password: "),
    host="127.0.0.1",
    port=3306,
    database="medimetrics"
)

engine = create_engine(url)

# Load cleaned doctors dataset
doctors = pd.read_csv("../data/processed/cleaned_doctors.csv")

print("Doctors dataset loaded:", doctors.shape)

# Import doctors into MySQL
doctors.to_sql(
    name="doctors",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=100
)

print("Doctors table successfully imported!")
```
